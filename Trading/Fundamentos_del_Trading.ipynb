# Historia

1. Historia Lineal

   + Mesopotamia:
       + Quedaba en lo que hoy conocemos como: Irak, Turquia y Siria. Es el lugar donde todo empez.
    
       + 2250-2000 BC: inicia la renta de las tierras.
         
       + Los sumerios manejan las tierras, hay superavit de tierras con acceso a aguas.
         
       + 2000 BC: la tercera dinastia Ur recibe impuestos (gun-mada) en forma de renta, remanentes de la produccion obtenida.
         
       + De no haber remanentes se iba a la guerra entre vecinos o se optaba por adquirir prestamos (usura), dando origen a los intereses.
         
       + 1754 BC: surge el Codigo de Hammurabi con las primeras leyes contractuales y de comercio, se habla ya de deudas y se amplia la practica de la usura.
    
   
   + Grecia
       + En el siglo IV AC, en la antigua Grecia surge el libro "Oeconomicus" donde se hacen preguntas para llegar a conclusiones propias con relacion a la contratacion de granjeros (no esclavitud) y surgen los primeros 'manager' o administradores de tierras.
    

       + Surge otro libro: "Alegorias de la granja, de Grecia a Roma" donde se inicia a comprar tierras con problemas, para despues trabajarlas y las vendian con alguien interesado (asi surgen los margenes de ganancia). Nota: House flipping: Es la actividad de comprar propiedades arruinadas, se trabajan y se mejoran, para luego venderlas con un margen atractivo (lo que hoy conocemos como inversionistas de bienes raíces).
    

    + Roma (ya una república bien establecida)
        + En el año 200 AC, se consolidad la inversion en bienes raices, haciendo ricas a muchas familias de la elite romana.

2. Historia del `interes`
   + El interes es el motivo por el que una persona o una institucion esta interesada en prestarte dinero
      
   + En egipto 300 BC, las tasas de interes eran del 100%, no fue hasta Ptolomeo (post 30BC) que las tasas de interes se igualaron con las del naciente imperio Romano en un 25%

   +  Roma fue financiando mucho de su expansion, gracias al movimi9ento prestamo-interes de la clase noble para financiar campañas de conquista
  
   +  Durante la edad Media y el Renacimiento, el fenomeno usurero (prestamos con intereses muy altos) gano popularidad clandestina, pero tambien perdio buena reputacion
  
   +  De todas las categorias o tipos de inversion el comercio (trade) era el mas riesgoso y a la vez el que ams retorno podia ofrecer
  
   +  La figura de la corporacion surgio como una entidad que protegia a los inversionistas que pagaban la expedicion, esta acuataba como escudo.
  
   +  La primera corporacion que existio era The British East India Company que fue fundada en 1600 y cerro sus puertas despues de 1800, fue una de las pioneras en la entidad corporativa
  
   +  Esta empresa fue la primera que constituyo el modelo de ownership (propiedad privada) a traves de acciones (shares of stock) y la primera que produjo estructuras administrativas y legales como las de hoy en dia
  
   +  The British East India Company fue tambien la primera en usar un sistema de proteccion contra responsabilidad legal, protegiendo a sus inversionistas. Asi nacio la responsabilidad limitada (Limited Liability)
  
   +  Las corporaciones se convirtieron en el vehiculo ideal para invertir y financiar en las riesgosas expediciones
  
   +  En 1602, nacio The Dutch East India Company, esta se convirtio en la primera empresa "publica", poniendo a disposicion del publico inversionista sus acciones en el Amsterdam Stock Exchange
  
3. Breve historia de la inversion
   
   + En un principio la compra-venta de acciones publicas en los primeros Stock Exchange de Europa (Amsterdam, Belgica) sucedia en los cafes y en algunas plazas publicas
  
   + La primera bolsa de valores que emitio de manera formal acciones al publico similar a como se hacia con emisiones de deuda de gobierno fue la bolsa de valores de Londres
  
   + El London Stock Exchange fue creado en 1801 y rapidamente el New York Stock Exchange fue inaugurado en 1817 y se convirtio en la Bolsa mas importante y poderosa del Mundo
  
   + Las Bolsas de Valores del mundo encuentran su razon de existir en el concepto de 2 temas: el manejo de riesgos y como la corporacion es una entidad que nacio para protegerse del riesgo, y dos la busqueda de desarrollo economico y de negocios.
   

# Categorias y Tipos de inversion

+ Bienes inmuebles (Real State)
+ Productos Bancarios (Bank Products)
+ Bonos (Bonds)
+ Acciones (Stocks, ETFs)
+ Fondos de inversion y de Retiro / Pensiones (Mutual Funds, Pension / Retirement Funds)

+ Otros productos (Alternative Investments)



1. Bienes Inmuebles(Real state): Inversiones en tierra o en los edificion construidos sobre ella y/o el ecosistema natural habitado dentro de sus limites

   
   + Puede ser comercial o Residencial (Comprando propiedades o invirtiendo en instrumentos relacionados)
  
   + Se genera valor por medio de:
     
     *  Cash-flow / renueve con rentas (arriendo)
       
     *  Apreciacion del inmueble o accion sobre el paso del tiempo
       
   + Los factores locales/geograficos pueden afectar de manera positiva o negativa y tambien afecta el mercado del sitio

2. Productos bancarios(Bank Products): Productos adquiridos de la banca. Son generalmente liquidos (facilidad con la que un activo puede ser convertido en dinero efectivo sin afectar significativamente su valor), pero ofrecen un interes muy bajo
   
   + Algunos son:
  
     * Cuentas de ahorro: Alta liquidez, bajo costo pero bajo interes

     * Certificados de deposito: Mayor interes que las cuentas de ahorro pero menos liquidos y tienen un plazo de tiempo determinado
    
     * Cuentas de Mercado de Dinero: Cuentas de deposito similar a la cuenta de ahorro pero con mayor interes y generalmente con mayores restricciones administrativas

3. Bonos(Bonds): Instrumento de deuda donde un inversionista le hace un prestamo a una institución publica/agencia federal/institucion privada, a cambio de un pago de interes en un cierto plazo mas el repago del capital a la fecha de vencimiento del banco


+ 

+ Acciones (Stocks, ETFs): Son las inversiones que representan la propiedad proporcional de una entidad corporativa o empresa publica. Las acciones son la division de titulos de propiedad de una entidad corporativa o empresa publica

+ La manera en la que los inversionistas obtienen su retorno de inversion (o una perdida) es a traves de dos medios:

  * Dividendos de la empresa en caso de que los distribuya (los dividendos son una parte de las ganancias totales,a cada inversionista o socio se le da porcion de un porcentaje de una parte de las ganancias de la empresa)
 
  * Apreciacion o Despreciacion de el valor de cada accion en el mercado tras una transaccion de venta o compra (Capital Gains)

+ El valor de la accion de una empresa depende de varios factores que hacen que la inversion en el mercado bursatil de acciones (en cualquier parte del mundo) sea uno de los instrumentos mas volatiles y riesgosos:

  * Ambientes de mercado en general (politica, economia, ambiente social)
    
  * Ganancias de la empresa o del sector (ETFs) al que pertenece la empresa
    
  * Demanda o "Hype" del sector o de la empresa (IPO, FANG)
    
  * Factores "irracionales" o completamente random (un tweet de su CEO, un discurso de un funcionario publico de gran importancia internacional)
 
+ Fondos de inversion y Retiro: Vehiculo de inversion en el cual se agrupa a varios inversionistas con un fin o estrategia de inversion especifica y generalmente dejando su administracion y diversificacion en manos de un "Fund Manager"

+ Existen gran variedad de tipos de fondo de inversion:

  * Mutual Funds (Stocks, Bonds, Money Market, Open/Close, etc)
    
  * Exchange Traded Funds (ETFs)
    
  * Venture Capital Funds
    
  * Fondos de Fondos (Fondos Soberanos)
    
  * Fondos de Pension (Fondos de Retiro)
    
  * Hedge Funds (no registrados ante la SEC en Estados Unidos)

+ Insurance(Seguros de Vida u otros tipos de seguros): aunque no sea comun considerarlos como una inversion, su popularidad esta aumentando como estrategia de inversion.

- Ventajas:
  - Construye valor sobre el Premium (interes compuesto) con ventajas fiscales
  - Es una buena estrategia de "Estate Planning" (Planificacion Patrimonial)
  - Los Premium nunca incrementan
  - Proteccion tradicional contra Muerte inesperada
  - Sus dividendos pueden cubrir el Premium en cierto plazo
  - Obliga a ahorrar

- Desventajas:
  - Premium considerablemente mas alto que el Seguro de Vida de Termino (Term)
  - Poco flexible (La estructura de pago de premiums y las condiciones no cambian)
  - La compania que lo emite (Aseguradora) tiene completo control sobre el Premium

+ Derivados: Instrumentos de inversion cuto valor esta basado en el precio o valor de otro activo subyacente

+ Algunos son:

  * SWAP (Intercambios de cahs-flow, de intereses / deuda, moneda y precios flotantes)
  * Opciones (Contratos de derecho de compra-venta sobre stocks, basicos, etc)
  * Futuros (Contratos de compra o venta futura de un activo subyacente)
  * CFDs (Contrato por Diferencia de compra o venta donde el que compra/vende -segun si el valor se aprecia o se desprecia- paga la diferencia de precio del activo subyacente al vencimiento de este)
 
- Vehiculos alternativos de inversion: son mas complejos y poco comunes. Ayudan a la diversificacion de inversionistas sofisticados

- Algunos ejemplos son:

  - Notas Convertibles, Notas Reversibles, Safe Notes (Usados por Angels/Startups)
  - Mutual Funds Alternativos (Alt-Mutual Funds, con inversiones poco tradicionales)
  - Notas de Deuda con Principal Protegido
  - Inverse Exchange Traded Funds
  - Bonos de Catastrofe y otros vehiculos relacionados a eventos azarosos
  - ICOs y Criptomonedas (Bitcoin, etc)

# Traders, Objetivo y Estrategias
+ El mejor amigo del trader no es el tiempo, es el movimiento de accion del precio

+ El objetivo de un trader es generar valor maximizando su ganancia sobre la apreciacion de capital de la accion, gracias a los movimientos en el precio, generalmente en corto o mediano plazo

+ El largo plazo no es tan atractivo para un trader, a menos que sea para diversificar una estrategia (trend following)

+ Gracias a el movimiento de Accion de Precio, existe la Volatilidad

+ El inversionista deposita toda su confianza en el tiempo, el trader se enfoca en el movimiento del precio de una accion

+ Un trader no se fija en una temporalidad, un trader no esta estudiando el tiempo, el trader esta enfocado 100% en el movimiento de Accion de Precio

+ Un mutual fund da un 6% anual

# El inversionista
+ Tiene como mejor amigo al Tiempo

+ El inversionista esta orientado a revisar ganancia en medianos o largos plazos

+ El valor de la inversion de un inversionista esta determinado en la empresa

+ El inversionista usa principalmente el Analisis Fundamental

+ Su administracion de riesgo es mas relajada, dibido a los medianos y largos plazos

+ Su retorno es a traves de ganancias de capitales obtenidos

  * Incremento de valor de la accion
  * Dividendos obtenidos pasivamente
  * Efector del interes compuesto (Acumulacion de capital)

+ El shorting es apostar en contra de una accion, un claro ejemplo de esto es que una moneda valla a perder valor y se haga shorting lo que generara una ganancia

# Empresa privada vs Empresa publica
## Empresa privada:
+ Propiedad individual, grupal o por instituciones privadas
+ Puede durar mucho tiempo o indefinidamente
+ Los owners son los que toman las decisiones de quien puede adquirir participacion en la empresa
+ LOs resultados financieros se mantienen cerrados y no son compartidos

* A las empresas privadas tambien se les conoce como Corporacion Cerrada
* Una empresa privada no tiene necesariamente menos recursos que una empresa publica
* Las empresas privadas emiten titulos de propiedad que no se pueden comprar publicamente, solo de manera privada
* No son propiedad del gobierno
* Tienen responsabilidad limitada, es decir, protegen a sus inversionistas, protegen a los duenos, y hacen un escudo legal que protege los activos de los duenos (como en las expediciones de The British East India Company)
* No publican resultados financieros

**LLC: Limited Liability Company (Compania de Responsabilidad Limitada). Tiene 3 posibilidades:**

* **Flow Through**: Un vehiculo para una persona, o para una asociacion entre dos o mas personas. Sirve como un escudo legal, pero no como una identidad independiente desde la perspectiva financiera y legal
* **Semi-Flow Through**: Simula una corporacion de tipo S
* **Entity Independent**: Simula una corporacion de tipo C. Hacer trading como individuo no es siempre la mejor opcion (hacer trading como empresa)


## Empresa publica:
+ Tiene la posibilidad de ser propiedad de sus fundadores y de la comunidad inversionista adquiriendo los titulos de propiedad de la empresa publica.
+ La propiedad publica cambia constantemente de manos, no es algo que se quede estatico, la propiedad publica es lo que hace que un mercado bursatil intercambie manos con transacciones compra-venta de acciones que suceden.
+ Cualquiera puede comprar acciones, pero las decisiones las toma el Consejo y los accionistas con derecho a a voto.
+ Los resultados financieros se deben mantener abiertos.

* Normalmente las empresas publicas comenzaron por empresas privadas
* Ademas de su estructura privada de acciones inicial, emiten y ofertan al publico inversionistas una cantidad determinada de acciones en los mercados bursatiles
* El mecanismo a traves del cual una empresa privada se vuelve publica se llama `Oferta Publica Inicial (IPO)`
* Una de las causas principales por las que una empresa publica lanza su IPO tiene que ver con financiamiento agresivo o estrategia de salida para inversores iniciales (Angeles, Fondo de Capital de Riesgo)
* Pueden retornar a ser empresas privadas
* Tienen Responsabilidad ante Autoridades Financieras, Regulatorias y Gubernamentales
* Estan obligados a publicar resultados financieros

# Conflicto Oferta-Demanda en trading de acciones
## Conflicto inicial entre dos partes:
+ El precio de oferta `"BID"`, es lo que el comprador esta dispuesto a pagar por la accion (Stock Price)

+ El precio `"ASK"` o de demanda, es lo que el vendedor esta dispuesto a aceptar que le paguen por la accion



+ Bid es el precio mas alto que el comprador esta dispuesto a pagar, y ask el precio mas bajo al que el vendedor esta dispuesto a vender

+ `"Spread"` es la distancia que hay entre el precio de oferta el el precio de demanda (Distancia entre Bid y ASK)

# Terminologia:
## Trader
 + El trader esta muy enfocado en el corto plazo, pueden ser minutos, horas, dias, semanas, o meses
   
 + El trader necesita un alto monitoreo
   
 + Se ocupa un analisis mas tecnico
   
 + Apreciacion de capital (copmpra barato, vende caro)
   
 + Un trader es mucho mas fragil al movimiento del mercado
   
 + Un trader no esta apostando, un trader especula. No es lo mismo hacer especulacion a apostar. Cualquier persona que hace negocios de alguna u otra manera hace una especulacion
   
 + Un trader puede sacarle provecho a ambas direcciones (long, short) del mercado (hacia arriba o hacia abajo)
   
 + Alta frecuencia, un trader esta abierto a entrar y salir, y esto hace que sea mas costoso
   
 + Accion de precio (MOMENTUM), el trader esta interesado en el monitoreo de accion de precio
   
 + El riego de un trader es `MEDIO/ALTO`
   
 + Hay una gran carga emocional y alto nivel de estres, un trader pierde dinero por la emocion, ser poco emocional ayuda a ser mejor trader
   
 + Hay una mayor carga fiscal, ya que hay un movimiento de ganancia de capital

## Inversionista

## Short
+ Un short espera ganar gracias a una caida en la accion de precio, veden alto para buscar comprar bajo
+ 
- **Nota:** Sell Short, apostar en contra. Pides un prestamo al Broker y compras una accion que crees que va a caer. Vendes y cuando la accion pierda dinero (segun el analisis tecnico), entonces compras esta accion mas barata y ganas la diferencia. En otras palabras, vendes una accion que no posees en el momento con la esperanza de comprarla a un precio mas bajo.

- **Proceso:**
  - Tras un analisis tecnico serio identifico que las acciones de X compania (que en ese momento cuestan USD 2 c/u) van a bajar a USD 1.
  - No poseo ninguna accion de dicha compania por lo que no me afecta economicamente.
  - Quiero aprovechar dicha situacion para ganar dinero.
  - Le pido prestado a mi broker 100 acciones (cantidad de referencia) e inmediatamente las vendo al precio del mercado que son USD 2 por lo que el broker no pierde dinero.
  - Me quedo con el dinero como parte del prestamo (USD 200).
  - Espero a que se cumpla lo establecido en el analisis tecnico (baja del precio de la accion de USD 2 a USD 1).
  - Compro 100 acciones (las que me habia prestado mi broker) al precio nuevo (USD 1) y entonces de los USD 200 que tenia solo gasto USD 100 y esa diferencia es mi ganancia.
  - Devuelvo las 100 acciones al broker mas comisiones porque me presto dichas acciones.
  - Repetir este ciclo tantas veces se identifique la oportunidad (opcional).

- **Observaciones:**
  - Se debe conocer los montos de las comisiones que cobra el broker por los diferentes servicios de forma que la transaccion valga la pena.
  - Si bien siempre existe la posibilidad de perder dinero, con el analisis tecnico lo que se hace es minimizar lo mas posible la posibilidad de perdida.  
    + Ejemplo referencial (no tomar como algo fijo): sin analisis 100% de posibilidad de perdida, con analisis +90% de posibilidad de ganancia.

# Analisis fundamental
+ Es un metodo de evaluacion cualitativa y cuantitativa de un activo o inversion (en este caso una stock) para determinar su valor intrinseco real actual y futuro.
## Los factores del analisis fundamental
+ La `industria` a la que pertenece

+ La `economia` en general, tanto mundial, como nacional (en case de que este localizada en cierto mercado especifico)

+ Los `resultados` de ventas y el nivel de performance que esa empresa pueda entregar

+ La `valuacion` “real” (valor intrinseco) que tanto de manera interna como de manera externa (analistas, organismos de manera independiente que tienen a X compania como linea de negocios en hacer una valuacion responsable o por lo menos referente de cuanto podria valer en realidad de acuerdo a su deuda, a sus ingresos, a sus activos, etc.).

**Nota 1:**
- El valor intrinseco de una compania es el valor mas cercano a la realidad (sin importar la especulacion que pueda existir y sin dejarse llevar por otros factores externos) al valuar una compania (tanto actual como futuro).

**Nota 2:**
- La combinacion de los factores va a indicar en conjunto con los resultados pasados una expectativa sobre: “Creemos que podemos llegar a este resultado X, Y o Z, etc.”.
- Estos tres, cuatro o cinco o mas factores van a poder influir y determinar si una compania va en el camino correcto de seguir creciendo.

- Hay muchos factores que influyen en el comportamiento de accion de un precio.
    * Por ejemplo, un precio de accion de una empresa puede estar sobrevaluado y aunque los resultados financieros de dicha empresa sean positivos, es probable que el precio de la accion caiga.

# Analisis Tecnico
### Velas Japonesas (Japanese Candle Chart)
+ No hay analsis tecnico si no hay una comprension de la fluctuacion del precio. El movimiento de accion de precio es en si mismo el analisis tecnico. Por supuesto, comprender las velas japonesas (Japanese Candle Chart) es fundamental para realizar analisis tecnico

+ Las variaciones de precio en las graficas son producidas por el `desacuerdo de precio entre compradores y vendedores`, cuando estos dos logran llegar a un `acuerdo de precio`, se produce la transaccion que revela una diferencia de valor. En palabras mas sencillas, el comprador accede a X precio por que piensa que podria valer mas, en cambio el vendedor accede por que piensa que podria valer menos.

## Partes de la vela japonesa

### Que representan las velas japonesas
+ las velas son medida de tiempo, esto quiere decir que puede representar: una semana, un mes, una semana, un día, 4 horas, 1 hora, 30 min, 15 min, 1 min, etc.

+ Las lineas que sobresalen de la vela son llamadas sombras y en sus extremos marcan el precio mas alto, y el precio mas bajo al que llegaron en la medida de tiempo seleccionada y analizada (como se ve en la imagen anterior, de las partes de las japanese cnadle chart)

+ E.G
  
  * Si la medida de tiempo fura 1 dia, las punta de arriba de la sombra seria el precio mas alto alcanzado ese dia y la punta de abajo de la sombra seria el precio mas bajo alcanzado ese dia, esto tanto en la vela verde como en la vela roja

+ El cuerpo coloreado de verde es el "cuerpo de la vela" y muestra, cual fue el valor incial en su extremo inferior (open) y cual fue el valor al final en su extremo superior (close)

+ En el caso de la vela roja, el precio en el que comenzo en su extremo superior (open) y el precio de cierre en su extremo inferior (close)

+ 

+ Una vela verde representa que el precio termino mas alto de lo que inicio
  
+ Una vela roja representa que el precio termino mas abajo de lo que inicio
  
+ Las velas te indican 4 momentos:
  * Donde abrio el precio.
  * Donde cerro el precio.
  * El momento mas alto al que llego el precio.
  * El momento mas bajo al que llego el precio.

### Hay distintos tipos de velas verdes (bull):

+ Estas son algunas:

+ Primera:
  * donde abrio es el punto mas bajo de esa temporalidad
  * donde cerro es el punto mas alto de esam temporalidad
  * Esta vela es muy bullish, un optimismo muy grande

+ Segunda:
  * El mercado abrio en la parte baja de donde comienza la base verde de la vela
  * cerro en la parte superior del cuerpo de la vela
  * durante la temporalidad hubo un desacuerdo entre vendedores y compradores donde el precio estuvo por debajo del que abrio
  * donde cerro el precio fue el punto mas alto, esta vela no es tan optimista como la primera, pero si muestra un optimismo fuerte

+ tercera:
  *  termina el dia mucho mas alto de lo que comenzo
  *  hubo momentos en el dia en el que el precio estuvo en un punto mas alto de donde cerro
  *  tambien hubo momentos en el dia en el cual el precio estuvo mas abajo de donde abrio
  *  Esta vela ya muestra un equilibrio de precio entre compradores y vendedores

+ cuarta:
  * la distancia de donde abrio el precio y donde cerro, es muchisimo mas corta que la anterior
  * esta vela representa un cambio de parecer de direccion en el mercado, probablemente cambia de ser una opinion alcesta a una opinion bajista

+ quinta:
  * El precio abre en la parte mas baja
  * durante el dia no hay un punto mas bajo que de donde salio
  * hay un intento de navegacion hacia arriba, y baja de manera tremenda a la parte inferior del acuerdo entre vendedores y compradores

+ 

### Hay distintos tipos de velas rojas (bear):

+ Algunas son:

+ Primera:
  * En este tipo de vela el precio solamente bajo, nunca estuvo por encima de donde abrio. Donde cerro fue el punto mas bajo (una vela muy pesimista)


+ Segunda:
  * El precio de apertura y cierre es cercano, el precio termina por debajo de donde abrio (no hay tanto pesimismo, pero sigue habiendo ya que el precio estuvo muy por encima de donde abrio, pero no se puedo concretar en esa zona entonces bajo)

+ Tercera:
  * Existe un balance entre las fuerzas compradoras y las fuerzas vendedoras, es una vela equilibrada (neutral)

+ Cuarta:
  * Esta vela muestra que la distancia de la fluctuacion del precio es mucho mas grande que el cuepo de la vela, se comienza a ver una reversion y hay muchisimos compradores preparandose para empezar a comprar

+ Quinta:
  * El precio de cierre esta muy alejado de la fluctuacion mas baja del precio, es mas notable el intento de fuerza a que el mercado cambie de direccion

### Diferencias entre las velas Bull and Bear
+  Bear (vela roja): ganan compradores, esto ya que compran mas barato, y no gastan tanto capital (dinero) en las adquisiciones
+  Bull (vela verde): ganan vendedores, esto ya que venden mas caro, lo que hace que obtengan mas capital o un mayor porcentaje de retorno en la venta

# Moving average
## ¿Que es el moving average?  
- Es una herramienta de analisis tecnico que ayuda a identificar tendencias analizando periodos de un año (en dias).  
- Se usan lineas de promedio de 8, 21, 34 y 89 dias anteriores.  
- Si un movimiento de accion de precio rompe esos promedios es indicativo de un cambio de tendencia hacia arriba o hacia abajo.  

### Nota  
- Se recomienda que el precio atraviese todas o muchas de las lineas de Moving Average en un grafico, pero no siempre significa que vaya a darse un cambio de tendencia.  
- Habria que considerar otros factores.  

### Uso de varios indicadores  
- Se deben usar varios indicadores para sacar conclusiones sumando los resultados de todos.  
- Ejemplo: lectura de velas japonesas + Moving Average + otros 2 indicadores.  
- Si 3 de ellos sugieren que la tendencia es alcista al 90%, se apuesta en **long**.  
- Si es lo contrario, se apuesta en **short**.

# Que es el indicador MACD (Moving Average Corvengence Divergence) y cual es su teoria
+  El MACD es un indicador el cual se utiliza para el analisis tecnico, para evaluar posibles señales de compra o venta en el mercado de valores. El MACD se calcula restando un promedio de 26 dias de otro de 12 dias. Ademas de estas dos "lineas MACD", se agrega una "linea de señal" que se situa a 9 dias. Cuando la linea azul (MACD) cruza hacia arriba la linea amarilla (Señal), en teoria indica una señal de compra (a veces funciona otras veces no), hay señales falsas y verdaderas

## Como se debe utilizar el MACD
+ El MACD es una muy buena herramienta para usarla en una temporalidad en dias (estrategia a largo plazo). No es recomendable para day trading o trading a corto plazo, ya que la señal puede llegar tarde, despues de que el movimiento significativo del precio ya ocurrio. Muchos inversores consideran este indicador como un recurso poco confiable por si solo debido a su naturaleza lenta, pero puede ser util cuando se utiliza junto con otros indicadores basicos.


# Puntos importantes para poner

- **Take Profit (TP)**: orden que se coloca para cerrar automaticamente una operacion cuando el precio llega a un nivel predeterminado de ganancia.  
- **Stop Loss (SL)**: orden que limita la perdida cerrando automaticamente la operacion cuando el precio alcanza un nivel especificado en contra de la posicion.  
- **Target**: nivel de precio objetivo al que se espera que llegue el mercado, usualmente coincide con el take profit planeado.  
- **Entry**: punto o nivel de precio en el que se decide abrir una operacion (compra o venta).
