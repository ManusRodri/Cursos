# Procesos de arranque y encendido de dispositivos:

+ Al prender tu computadora o dispositivo movil, una señal electrica viaja hacia la `tarjeta madre`, esta electricidad, que es una onda, el dispositivo lo interprera en pulsos altos y pulsos bajos, un pulso alto es 1 y un pulso bajo es 0, a esto se le conoce como `los Bits` -> unidad minima de informacion.



+ Esta señal se envia hacia la tarjeta madre, la cual tiene un chip especial de arranque que recibe la señal, en este chip se encuentra un sistema fundamental llamado:

    * `BIOS` (Basic Input Output System) -> nombre antiguo.
    * `UEFI` (Unified Extensible Firmware Interface) -> nombre nuevo

+ En dispositivos moviles, como celulares, el chip encargado es el `Boot Loader`, y el proceso que ejecuta es diferente acorde al OS (Operative System)

    * Andriod -> `PBL` (Primary Boot Loader)
    * Iphone -> Secure Room, este luego carga el iBoot

+ Estos chips reciben la señal electrica y ejecutan un procedimiento interno, el cual esta grabado en el hardware de los transistores del chip. Este proceso se llama `POST` (Power On Self Test), este proceso lo que hace es que comprueba que todo en la computadora este funcionando correctamente y este conectado.

+ Cualquier sonido inusual al incio indica un fallo detectado por el POST. Si todo esta en orden, lo siguiente a proceder es buscar en la unidad de memoria permanente Las instrucciones de arranque del OS.

+ El codigo esta organizado en ceros `(0)` y unos `(1)`, con un lenguaje que lo entiende la `CPU` (Central Processing Unit), este componente es le encargado de realizar todos los calculos matematicos que se realizan para poder generar: imagenes, sonido, videos o procesos fundamentales o acciones que uno quiera realizar.

+ En computadoras o en dispositivos de gran tamaño, la CPU es un chip que se puede cambiar y es de gran tamaño, algunos ejemplos de chips son:

    * Intel Core Ultra 9 285K
    * AMD Ryzen 9950X3D

+ En celulares o dispositivos con un tamaño reducido, los chips se encuentran soldados y no se pueden cambiar, alrededor integra otros componentes de ese chip, a esto se le conoce como: `SoC` (System on a Chip), algunos ejemplos de chips son:

    * Qualcomm Snapdragon 8 Gen 3
    * Apple A19 Pro
    * Google Tensor G5

+ El lenguaje de ceros y unos, que sale de la unidad de memoria permanente y llegas a la CPU, instrucciones las cuales inician a ejecutar diferentes procesos, esas instrucciones nativas que interpreta la CPU, es un lenguaje que se conoce como `Assembler`.

+ La CPU ejecuta el `codigo de arranque del sistema operativo`, enviando las estructuras de informacion de los componentes del dispositivo y el uso de programas o aplicaciones a una memoria temporal de muy alta velocidad, que se conoce como `Memoria RAM` (Random Access Memory), esas estructuras se van aramando, como si se estuviara armando una ciudad desde cero a partir de sus planos.

+ El corazon del OS se conoce como `Kernel`, es el codigo en Software qe interpreta la CPU y estructura la memoria RAM para funcionar como un intermediario entre el `Software` y el `Hardware`, cada vez que tu haces una accion, el Kernel traduce las ordenes dadas por esa accion, para que esas acciones pasen de la computadora a la realidad.

+ El Kernel luego arranaca los procesos de seguridad del OS, de esa forma es posible que una computadora tenga datos con multiples dueños, pero que solo puedan se accedidos por su dueño, este es el `sistema de Usuarios y Contraseñas` que crea llaves de acceso que descifran los datos privados del usuario

+ El Kernel tambien busca en los archivos del OS, el proceso de arranque de las pantallas, las pantallas son matrices de puntos luminosos llamados `Pixeles`.

+ El manual de los elementos integrados, como los son: mause, teclado, pantalla, wifi, trackpad, etc. Este tipo de manual se expresa en software  y ese software se conoce como `Drivers`.

+ Al ingresar tu usuario y contraseña y le das enter, eso envia una señal electrica a la CPU, la CPU ecuentra la orden del OS de ir a buscar en la unidad de memoria permanente un `proceso criptografico matematico` donde esa contraseña descifra la `llave de acceso` al Kernel a tus datos privados.

# Como funciona un circuito electronico

+ Los circuitos electronicos son la base de la tecnologia moderna, estos funcionan controlando el flujo de electricicdad, que es el movimiento de electrones en un medio conductor (un cable), el flujo de la electricidad es muy parecido al del agua, y este fluye desde el `Polo negativo (-)` hacia el `Polo positivo (+)`.

+ El voltaje → es como la presion de agua en un tubo (`Tension o Fuerza`), se mide en `Voltios (V)`

+ El amperaje → es como el caudal del agua, es decir la cantidad de agua que pasa (`Intensidad de la corriente`), se mide en `Amperios (A)`

+ El vatio → es la energia total entregada por el agua (`Potencia`), se mine en `Vatios (W)`

+ El ohmnio → es la resistencia generada por el canal sobre el agua (`Friccion o Resistencia electrica`), se mide en `Ohmios (Ω)`

## Componentes Electricos

+ La electricidad, se puede convertir en luz, por medio de una pieza electronica llamada `LED`, pero estos LEDs se pueden calentar y quemar si reciben demasiada corriente. Para controlar y reducir el voltaje que una pieza electronica le ponemos antes del flujo un componente llamado `Resistencia`. La resistencia reduce el amperaje y absorbe esa energia y la libera como calor.



+ Motores electricos: convierten la energia electrica en movimiento mediante el uso de imanes que reaccionan al flujo electrico. Existen diferentes tipos:

    * Motores estandar: utilizados en electrodomesticos
    * Motores de precision (`Servos`): motores empleados en brazos roboticos o en equipos de alta presicion
    * Motores de alta potencia: motores utilizados en vehiculos pesados, como los Teslas

## ¿Como se relacionan los circuitos con la computación moderna?

+ Un elemento usado frecuentemente en los dispositivos, es el `Transistor`:ese pequeño elemento es el encargado de enecender o apagar otros componentes segun ciertas condiciones, es decir un interruptor electronico.

+ La computacion moderna se basa fundamentalmente en circuitos electronicos avanzados. Los `procesadores o microchips` son en esencia combinaciones de millones de transistores que realizan las operaciones matematicas.

+ Para almacenar los resultados de estas operaciones, las computadoras utilizan diferentes tipos de memoria:

    * `Condensadore`s: Componentes que almacenan electricidad temporalmente mientras hay flujo electrico continuo.

    * `Memoria RAM`: Formada por grupos de condensadores y transistores, almacena informacion de manera temporal mientras el dispositivo se encuentre prendido

    * `Discos duros (HDD)`: Utilizan reacciones fisicas magneticas para guardar informacion de forma permanente, incluso sin electricidad

    *`Discos de estado solido (SSD)`: grupo de condensadores y transisores de puerta flotante para guardar cargas electricas que representan 0s y 1s, alamcenando datos sin necesidad de electricidad

+ En estas memorias, un condensador lleno de electricidad representa un 1, mientras que uno vacio representa 0. Estos bits son la base de la representacionm digital de la informacion y se organizan en estructuras que los procesadores interpretan como instrucciones de software

+ EL software o codigo de programacion almacenado en la memoria, viaja al procesador a traves del circuito para ser ejecutado, convirtiendo esos 1s y 0s en aplicaciones y sistemas que se utilizan a diario

+ Los circuitos electronicos representan la base fundamental de toda nuestra tecnologia actual, desde los dispositivos mas simples hasta los mas complejos. Comprender sus principios basicos nos permite apreciar mejor el funcionamiento interno de los aparatos que utilizamos cotidianamente y nos abre las puertas a un mundo fascinante de posibilidades tecnologicas.

# Que es un bit y que es un byte

### Comunicacion por medio de luz

+ La luz es una `onda` que fluctua y dependiendo de su amplitud se hace visible o no, nos puede ayudar a transmitir mensajes.

    

+ El mapa de las de las amplitudes de onda se le conoce como `espectro electromagnetico` 

+ Las `ondas de radio` (Wifi, Bluetooth, Celulares), tambien es luz, pero no perceptible para el ojo humano.

+ Las ondas de radio pueden atravesar objetos fisicos con facilidad gracias a su amplitud de onda.

### Conversion de ondas en sonido

+ Las antenas de radio captan las señales de radio y por medio de un `circuito rectenna` convierte las señales en electricidad.

+ Esa electricidad al mover una membrana genera sonido.

### Comunicacion binaria por medio de luz

+ Una luz encendida representa un 1 y una luz apagada un 0, (bit).

+ Para transmitir mensajes, previamente se debe acordar el tiempo de duracion de cada bit.

+ En circuitos electronicos, el 1 se representa con un alto voltaje y el 0 con un voltaje nulo o bajo.

### Evolucion de la transmision de mensajes

+ El Telegrafo usaba tonos largos y cortos (codigo morse), para transmitir mensajes.

+ Las computadoras usan voltajer y bits para representar datos.

### Tala ASCII y representaciones de caracter

+ ASCII asocia numeros con caracteres.

+ En computadoras los caracteres se almacenan como numeros binarios.

### Sistemas de numeracion

+ Base 10: sistema decimal (0-9)

+ Base 16: sistema hexadecimal (0-9 y A-F)

+ Base 2: sistema binario (1-2)

### Representacion de numeros en binario

+ Se usan posiciones para duplicar su valor: 1, 2, 4, 8, 16, 32, 64, 128, ...

+ Un 1 en esa posicion suma ese valor; un 0 no suma

+ 8 posiciones binarias (`Bits`), forman un `Byte`

+ Un byte permite representar valores desde 0 hasta al 255

### Byte y su uso en la computacion

+ `1 Byte = 8 bits`

+ Los archivos utilizan bytes para representar caracteres, saltos de lineas e instrucciones.

+ Tambien es usualmente usado para calculos matematicos y procesos logicos que se envian al procesador y son ejecutados por Assembler.

+ Tambien se usa para representar simbolos con ASCII.

### Aplicaciones practicas de los Bytes

+ Imagenes -> colores

+ Emojis -> requiere mas bits (UTF-8 a UTF-32), (UTF -> Unicode Transformation Format)

+ Textos

### Unicode y compatibilidad internacional

+ El Unicode Consortium regula la equivalencia entre Bytes y simbolos de todo el mundo.

### Programacion y traduccion al leguaje maquina

+ Los lenguajes de programacion se traducen a assembler (Bytes)

+ Los compiladores son los que realizan esta conversacion entre el programa y assembler.
